```markdown
## Étape : Prétraitement des données (Data Preprocessing) & Pipeline
```

````python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.base import BaseEstimator, TransformerMixin

# 1. Vérifier d'abord la présence de valeurs manquantes dans X et Y
print("Valeurs manquantes dans X :")
display(X.isnull().sum().to_frame(name='Manquantes (X)'))
print("\nValeurs manquantes dans Y :")
print(f"Nombre de valeurs manquantes dans Y : {Y.isnull().sum()}\n")

# 2. Classe personnalisée pour transformer la variable 'Date' en caractéristiques numériques
class DateFeaturesExtractor(BaseEstimator, TransformerMixin):
    def fit(self, X_df, y=None):
        return self

    def transform(self, X_df):
        # Copie pour ne pas modifier l'original
        df_out = X_df.copy()
        # S'assurer que la colonne est au format datetime
        date_series = pd.to_datetime(df_out['Date'])

        # Extraction des caractéristiques numériques
        df_out['Date_Month'] = date_series.dt.month
        df_out['Date_Day'] = date_series.dt.day
        df_out['Date_DayOfWeek'] = date_series.dt.dayofweek

        # Supprimer la colonne d'origine 'Date' après extraction
        df_out = df_out.drop(columns=['Date'])
        return df_out

# 3. Définition des groupes de caractéristiques générées après extraction de date
numerical_features = ['Unit Price (USD)', 'Date_Month', 'Date_Day', 'Date_DayOfWeek']
categorical_features = ['Product', 'Category', 'Region']

# 4. Construction du ColumnTransformer pour le scaling et l'encodage
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features)
    ]
)

# 5. Pipeline global qui intègre d'abord l'extraction de dates, puis la transformation des colonnes
preprocessing_pipeline = Pipeline(steps=[
    ('date_extractor', DateFeaturesExtractor()),
    ('col_transformer', preprocessor)
])

# 6. Afficher les transformations prévues
print("Structure du pipeline de prétraitement :")
print(preprocessing_pipeline)
````

```markdown
### Explications des étapes :

1. **Vérification des données manquantes** : Nous confirmons qu'il n'y a pas de valeurs manquantes dans $X$ ou $Y$ pour éviter des erreurs lors de l'application des modèles.
2. **Extraction temporelle (`DateFeaturesExtractor`)** : La variable `Date` est convertie en trois variables numériques (`Date_Month`, `Date_Day`, `Date_DayOfWeek`) pour que le modèle puisse en capter la saisonnalité.
3. **Encodage One-Hot** : Les colonnes textuelles (`Product`, `Category`, `Region`) sont transformées en vecteurs binaires (0 ou 1) avec `OneHotEncoder(handle_unknown='ignore')` pour gérer de nouvelles catégories potentielles.
4. **Standardisation (`StandardScaler`)** : Les variables numériques (le prix unitaire et nos nouvelles variables temporelles) sont centrées et réduites (moyenne de 0, écart-type de 1) pour homogénéiser les échelles.
5. **ColumnTransformer & Pipeline** : Ces étapes sont encapsulées dans un pipeline global. **Aucun ajustement (`.fit()`) n'est fait à ce stade** afin de respecter la règle anti-data leakage ; il sera entraîné uniquement sur l'ensemble d'entraînement à l'étape suivante.
```

```markdown
## Étape : Division des données (Train/Test Split)
```

````python
from sklearn.model_selection import train_test_split

# 1. Division initiale : 80% entraînement (full), 20% test
# Comme il s'agit d'une régression, nous n'utilisons pas le paramètre 'stratify'
X_train_full, X_test, Y_train_full, Y_test = train_test_split(
    X, Y,
    test_size=0.20,
    random_state=42
)

# 2. Deuxième configuration : Sélectionner 50% des données d'entraînement (40 observations)
# Nous utilisons à nouveau train_test_split sur l'ensemble d'entraînement complet pour garantir la reproductibilité
X_train_50, _, Y_train_50, _ = train_test_split(
    X_train_full, Y_train_full,
    train_size=0.50,
    random_state=42
)

# 3. Afficher les dimensions des différents ensembles
print("--- Configuration 1 : Entraînement Complet (100% de 80% = 80 obs) ---")
print(f"X_train_full : {X_train_full.shape}")
print(f"Y_train_full : {Y_train_full.shape}\n")

print("--- Configuration 2 : Entraînement Réduit (50% de 80% = 40 obs) ---")
print(f"X_train_50   : {X_train_50.shape}")
print(f"Y_train_50   : {Y_train_50.shape}\n")

print("--- Ensemble de Test Unique (Commun aux deux configurations = 20 obs) ---")
print(f"X_test       : {X_test.shape}")
print(f"Y_test       : {Y_test.shape}")
````

```markdown
### Explications des résultats :

- **Reproductibilité** : Le paramètre `random_state=42` garantit que la répartition des données reste identique à chaque exécution.
- **Ensemble de test commun** : Nous avons isolé en premier lieu 20 % des données pour le test (soit 20 observations pour `X_test` et `Y_test`). Cet ensemble ne changera pas, ce qui permettra de comparer équitablement nos deux futures expériences d'entraînement.
- **Les deux configurations** :
  - **Configuration 1 (Full)** : Utilise l'intégralité des données d'entraînement disponibles, soit 80 observations.
  - **Configuration 2 (50%)** : Extrait la moitié de l'ensemble d'entraînement précédent, soit 40 observations, pour simuler l'impact d'un volume de données plus restreint sur l'apprentissage des modèles.
```

```markdown
## Étape : Entraînement des Modèles de Régression (Configuration 100% vs 50%)
```

````python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# 1. Définition des estimateurs de base
lin_reg = LinearRegression()
tree_reg = DecisionTreeRegressor(random_state=42)
rf_reg = RandomForestRegressor(random_state=42)

# 2. Création des pipelines combinant le préprocesseur et les modèles
pipelines = {
    'Linear Regression': {
        'full': Pipeline(steps=[('preprocessor', preprocessing_pipeline), ('regressor', lin_reg)]),
        '50': Pipeline(steps=[('preprocessor', preprocessing_pipeline), ('regressor', lin_reg)])
    },
    'Decision Tree': {
        'full': Pipeline(steps=[('preprocessor', preprocessing_pipeline), ('regressor', tree_reg)]),
        '50': Pipeline(steps=[('preprocessor', preprocessing_pipeline), ('regressor', tree_reg)])
    },
    'Random Forest': {
        'full': Pipeline(steps=[('preprocessor', preprocessing_pipeline), ('regressor', rf_reg)]),
        '50': Pipeline(steps=[('preprocessor', preprocessing_pipeline), ('regressor', rf_reg)])
    }
}

# 3. Entraînement des 6 configurations de modèles
print("--- Début de l'entraînement des modèles ---\n")

# Configuration 1: Entraînement complet (80 observations)
print("[Configuration 1 - 100% (80 observations)]")
pipelines['Linear Regression']['full'].fit(X_train_full, Y_train_full)
print("- Linear Regression entraîné sur 80 observations.")
pipelines['Decision Tree']['full'].fit(X_train_full, Y_train_full)
print("- Decision Tree entraîné sur 80 observations.")
pipelines['Random Forest']['full'].fit(X_train_full, Y_train_full)
print("- Random Forest entraîné sur 80 observations.\n")

# Configuration 2: Entraînement partiel (40 observations)
print("[Configuration 2 - 50% (40 observations)]")
pipelines['Linear Regression']['50'].fit(X_train_50, Y_train_50)
print("- Linear Regression entraîné sur 40 observations.")
pipelines['Decision Tree']['50'].fit(X_train_50, Y_train_50)
print("- Decision Tree entraîné sur 40 observations.")
pipelines['Random Forest']['50'].fit(X_train_50, Y_train_50)
print("- Random Forest entraîné sur 40 observations.\n")

print("--- Tous les 6 entraînements sont terminés avec succès ! ---")
````

```markdown
### Explications de cette étape :

- **Encapsulation complète** : Chaque modèle est enveloppé dans un `Pipeline` qui inclut notre `preprocessing_pipeline`. Ainsi, l'extraction de la date, l'encodage One-Hot et la standardisation sont appliqués de manière isolée à chaque ensemble d'entraînement (`_full` et `_50`), garantissant l'absence de fuite d'information depuis l'ensemble de test.
- **Les 6 entraînements** : Nous avons ajusté les 3 algorithmes (Régression Linéaire, Arbre de Décision, Forêt Aléatoire) sur les deux volumes de données. Ces modèles entraînés sont maintenant prêts à être évalués et comparés sur les mêmes 20 observations de test (`X_test`, `Y_test`).
```

```markdown
## Étape : Évaluation Statistique des Modèles de Régression
```

````python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Initialisation des listes pour collecter les résultats des métriques
results_list = []
predictions_dict = {'Valeur Réelle (Y_test)': Y_test.values}

# Parcours de nos modèles et de leurs deux configurations d'entraînement
for model_name, configs in pipelines.items():
    for config_name, pipeline in configs.items():
        # 2. Faire les prédictions sur X_test
        preds = pipeline.predict(X_test)

        # Enregistrer les prédictions pour l'affichage comparatif
        label = f"{model_name} ({'100%' if config_name == 'full' else '50%'})"
        predictions_dict[label] = preds

        # 4. Calculer les métriques de régression
        mae = mean_absolute_error(Y_test, preds)
        mse = mean_squared_error(Y_test, preds)
        rmse = np.sqrt(mse)
        r2 = r2_score(Y_test, preds)

        results_list.append({
            'Modèle': model_name,
            'Données d\'Entraînement': '100% (80 obs)' if config_name == 'full' else '50% (40 obs)',
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'R²': r2
        })

# 5. Présenter les résultats dans un seul tableau comparatif
df_results = pd.DataFrame(results_list)
print("--- Tableau Comparatif des Performances des Modèles ---")
display(df_results)

# 6. Afficher également les valeurs réelles et prédites pour chaque modèle
df_predictions = pd.DataFrame(predictions_dict)
print("\n--- Comparaison des Valeurs Réelles et Prédites (20 observations de test) ---")
display(df_predictions)
````

````python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Liste pour stocker les résultats formatés
formatted_results = []

# Dictionnaire de correspondance des noms pour l'affichage demandé
model_display_names = {
    ('Linear Regression', 'full'): 'Linear Regression (100% Train)',
    ('Linear Regression', '50'): 'Linear Regression (50% Train)',
    ('Decision Tree', 'full'): 'Decision Tree Regressor (100% Train)',
    ('Decision Tree', '50'): 'Decision Tree Regressor (50% Train)',
    ('Random Forest', 'full'): 'Random Forest Regressor (100% Train)',
    ('Random Forest', '50'): 'Random Forest Regressor (50% Train)'
}

# Vérification et calcul des performances à partir des modèles en mémoire
for (model_key, config_key), display_name in model_display_names.items():
    if model_key in pipelines and config_key in pipelines[model_key]:
        pipeline = pipelines[model_key][config_key]
        # Prédictions sur l'ensemble de test commun de 20 observations
        preds = pipeline.predict(X_test)

        # Calcul des métriques demandées
        mae = mean_absolute_error(Y_test, preds)
        mse = mean_squared_error(Y_test, preds)
        rmse = np.sqrt(mse)
        r2 = r2_score(Y_test, preds)

        # Ajout des résultats arrondis à 4 décimales
        formatted_results.append({
            'Modèle': display_name,
            'MAE': round(mae, 4),
            'MSE': round(mse, 4),
            'RMSE': round(rmse, 4),
            'R²': round(r2, 4)
        })
    else:
        print(f"Attention : Le modèle {model_key} ({config_key}) n'est pas disponible en mémoire.")

# Conversion en DataFrame pandas et affichage
df_comparatif_final = pd.DataFrame(formatted_results)
display(df_comparatif_final)
````

```markdown
## Étape : Correction de l'isolement des modèles et Évaluation Finale
```

````python
from sklearn.base import clone
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Vérification de la taille des ensembles d'entraînement
print(f"Taille de X_train_full : {X_train_full.shape[0]} observations")
print(f"Taille de X_train_50   : {X_train_50.shape[0]} observations\n")

# 2. Utilisation de `clone` pour créer 6 pipelines structurellement identiques mais totalement indépendants
pipelines_correct = {
    'Linear Regression (100% Train)': clone(preprocessing_pipeline).set_params(col_transformer__num=StandardScaler()).set_output(transform="default"),
    'Linear Regression (50% Train)': clone(preprocessing_pipeline).set_params(col_transformer__num=StandardScaler()).set_output(transform="default"),
    'Decision Tree Regressor (100% Train)': clone(preprocessing_pipeline).set_params(col_transformer__num=StandardScaler()).set_output(transform="default"),
    'Decision Tree Regressor (50% Train)': clone(preprocessing_pipeline).set_params(col_transformer__num=StandardScaler()).set_output(transform="default"),
    'Random Forest Regressor (100% Train)': clone(preprocessing_pipeline).set_params(col_transformer__num=StandardScaler()).set_output(transform="default"),
    'Random Forest Regressor (50% Train)': clone(preprocessing_pipeline).set_params(col_transformer__num=StandardScaler()).set_output(transform="default")
}

# Recréer les pipelines complets avec leurs estimateurs respectifs
pipelines_correct['Linear Regression (100% Train)'] = Pipeline(steps=[('preprocessor', clone(preprocessing_pipeline)), ('regressor', LinearRegression())])
pipelines_correct['Linear Regression (50% Train)'] = Pipeline(steps=[('preprocessor', clone(preprocessing_pipeline)), ('regressor', LinearRegression())])
pipelines_correct['Decision Tree Regressor (100% Train)'] = Pipeline(steps=[('preprocessor', clone(preprocessing_pipeline)), ('regressor', DecisionTreeRegressor(random_state=42))])
pipelines_correct['Decision Tree Regressor (50% Train)'] = Pipeline(steps=[('preprocessor', clone(preprocessing_pipeline)), ('regressor', DecisionTreeRegressor(random_state=42))])
pipelines_correct['Random Forest Regressor (100% Train)'] = Pipeline(steps=[('preprocessor', clone(preprocessing_pipeline)), ('regressor', RandomForestRegressor(random_state=42))])
pipelines_correct['Random Forest Regressor (50% Train)'] = Pipeline(steps=[('preprocessor', clone(preprocessing_pipeline)), ('regressor', RandomForestRegressor(random_state=42))])

# 3 & 4. Entraînements indépendants
pipelines_correct['Linear Regression (100% Train)'].fit(X_train_full, Y_train_full)
pipelines_correct['Linear Regression (50% Train)'].fit(X_train_50, Y_train_50)

pipelines_correct['Decision Tree Regressor (100% Train)'].fit(X_train_full, Y_train_full)
pipelines_correct['Decision Tree Regressor (50% Train)'].fit(X_train_50, Y_train_50)

pipelines_correct['Random Forest Regressor (100% Train)'].fit(X_train_full, Y_train_full)
pipelines_correct['Random Forest Regressor (50% Train)'].fit(X_train_50, Y_train_50)

# 7, 8 & 9. Calcul et comparaison des nouvelles métriques
corrected_results = []
for name, pipeline in pipelines_correct.items():
    preds = pipeline.predict(X_test)

    mae = mean_absolute_error(Y_test, preds)
    mse = mean_squared_error(Y_test, preds)
    rmse = np.sqrt(mse)
    r2 = r2_score(Y_test, preds)

    corrected_results.append({
        'Modèle': name,
        'MAE': round(mae, 4),
        'MSE': round(mse, 4),
        'RMSE': round(rmse, 4),
        'R²': round(r2, 4)
    })

df_comparatif_corrigé = pd.DataFrame(corrected_results)
print("--- Nouveau Tableau Comparatif des Performances (Corrigé) ---")
display(df_comparatif_corrigé)
````

```markdown
### Explication des métriques et résultats en français simple :

- **MAE (Mean Absolute Error - Erreur absolue moyenne)** : Indique de combien d'unités de quantité (en moyenne) nos prédictions s'écartent de la réalité. Plus elle est proche de 0, meilleur est le modèle.
- **MSE (Mean Squared Error - Erreur quadratique moyenne)** : Calcule la moyenne des carrés des erreurs. Elle pénalise plus lourdement les grandes erreurs de prédiction.
- **RMSE (Root Mean Squared Error)** : C'est la racine carrée de la MSE, ce qui permet d'interpréter l'erreur dans la même unité que la variable cible (`Quantity`).
- **R² (Coefficient de détermination)** : Indique la proportion de la variance de `Quantity` expliquée par nos modèles. Un score de $1.0$ représente une prédiction parfaite, tandis qu'un score négatif indique que le modèle est moins performant qu'une simple moyenne.

*Remarque sur l'impact du volume de données :* En comparant les résultats entre la version `100% (80 obs)` et `50% (40 obs)`, vous pourrez identifier si la réduction du volume de données a dégradé la performance et quel modèle (linéaire, arbre ou forêt) s'est montré le plus stable sur ce petit échantillon.
```
